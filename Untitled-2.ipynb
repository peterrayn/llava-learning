```python
import torch
import torch.nn as nn
import torchvision
from torch.utils.data import DataLoader,Dataset
device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
from transformers import AutoTokenizer,AutoModelForCausalLM
from transformers import CLIPImageProcessor, CLIPVisionModel
from PIL import Image
import requests
import matplotlib.pyplot as plt


clip_vision_encoder = CLIPVisionModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
processor = CLIPImageProcessor.from_pretrained("openai/clip-vit-base-patch32")
tokenizer=AutoTokenizer.from_pretrained('distilgpt2')
gpt2=AutoModelForCausalLM.from_pretrained('distilgpt2').to(device)

tokenizer.pad_token=tokenizer.eos_token
```

```python
from datasets import load_dataset
dataset=load_dataset('jxie/flickr8k')
```

```python
import random
import re

def collate_fn(batch):
    # [{img,caption0-4}---{img,caption0-4}]
    img=[b['image'] for b in batch]
    img=processor(img)['pixel_values']
    img=torch.stack(img)
    text=["Describe the image:"+" "+b['caption_0'] for b in batch]
    text=tokenizer.encode(text,padding=True,return_tensors='pt')
    in_text=text[:,:-1]
    out_text=text[:,1:].clone()
    # print(out_text[0])
    out_text[:,:4]=-100
    out_text=torch.where(out_text!=tokenizer.pad_token_id,out_text,-100)
    return img,in_text,out_text
dataloader=DataLoader(dataset=train_dataset,batch_size=64,shuffle=True,collate_fn=collate_fn)
```

```python
# set false
for para in clip_vision_encoder.parameters():
    para.requires_grad=False
for para in gpt2.parameters():
    para.requires_grad=False
```

```python
llm_embed_token=gpt2.get_input_embeddings()
```

```python
class Projector(nn.Module):
    def __init__(self,input_dim,out_dim):
        super().__init__()
        self.linear=nn.Linear(input_dim,out_dim)
    def forward(self,x):
        return self.linear(x)
projector=Projector(768,768).to(device)
```

```python
# show the stupid effect

def see_effect(image,length=10,prompt='Describe the image:',):
# image = Image.open('./kitty.jpg')
    plt.imshow(image)
    input_img = processor(images=image, return_tensors="pt").pixel_values.to(device)
    input_text=tokenizer.encode(prompt,return_tensors='pt').to(device)
    print(input_text)
    print(tokenizer.decode(input_text))
    clip_vision_encoder.eval()
    gpt2.eval()
    with torch.no_grad():
        img_embed=clip_vision_encoder(input_img).last_hidden_state
        img_embed=projector(img_embed)
        for i in range(length):
            text_embed=llm_embed_token(input_text)
            combind_embed=torch.concatenate([img_embed,text_embed],dim=1)
            out=gpt2(inputs_embeds=combind_embed).logits
            predict_token=torch.argmax(out[:,-1:,:],dim=-1)
            input_text=torch.concatenate([input_text,predict_token],dim=-1)
    print(input_text)
    print(tokenizer.decode(input_text))
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.AdamW(projector.parameters(),lr=0.001)
```

```python
# see_before_train
see_effect(test_dataset[23]['image'],30)
```

```python
for epoch in range(1):
    for i,(img,in_text,out_text) in enumerate(dataloader):
        img,in_text,out_text=img.to(device),in_text.to(device),out_text.to(device)
        with torch.no_grad():
            img_embed=clip_vision_encoder(img).last_hidden_state
        img_embed=projector(img_embed)
        text_embed=llm_embed_token(in_text)
        combind_embed=torch.concatenate([img_embed,text_embed],dim=1)
        out=gpt2(inputs_embeds=combind_embed).logits
        loss=criterion(out[:,img_embed.shape[1]:,:].reshape(-1,tokenizer.vocab_size),out_text.reshape(-1))
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        if i>10:
            break
        print(f"{i}/{len(dataloader)}",loss.item())
```
