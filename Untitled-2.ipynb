```python
class Projector(nn.Module):
    def __init__(self,input_dim,out_dim):
        super().__init__()
        self.linear=nn.Linear(input_dim,out_dim)
    def forward(self,x):
        return self.linear(x)
```

```python
class LLM(nn.Module):
    def __init__(self,vocab_size,llm_dim,head_num,num_layer,context_length):
        super().__init__()
        encoder_layer=nn.TransformerEncoderLayer(d_model=llm_dim,nhead=head_num,batch_first=True)
        self.encoder=nn.TransformerEncoder(encoder_layer=encoder_layer,num_layers=num_layer)
        self.register_buffer("mask",torch.triu(torch.ones((context_length,context_length),dtype=torch.bool),diagonal=1))
        self.embed=nn.Embedding(vocab_size,llm_dim)
        self.posi_embed=nn.Embedding(context_length,llm_dim)
    def forward(self,vision,text):
        # B,L,D
        text=self.embed(text)
        x=torch.concatenate([vision,text],dim=1)
        position_embed=self.posi_embed(torch.arange(x.shape[1],device=x.device))
        x+=position_embed
        return self.encoder(x,mask=self.mask[:x.shape[1],:x.shape[1]])
```

```python
llm=LLM(vocab_size=tokenizer.vocab_size,llm_dim=256,head_num=4,num_layer=4,context_length=256)
projector=Projector(126,256)
```
